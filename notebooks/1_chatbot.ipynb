```python
#make a chat app using groq api using langchain
import os
from langchain_groq import ChatGroq
from dotenv import load_dotenv

load_dotenv()

llm = ChatGroq(
    model="openai/gpt-oss-120b",
    api_key=os.getenv("GROQ_API_KEY"),
    temperature=0
)
response=llm.invoke("Explain me ml in 3 line")
print(response.content)
```

```python
#make chart app using gemini api using langchain
import os
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv

load_dotenv()

model = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash",
    google_api_key=os.getenv("GOOGLE_API_KEY")
)
response=model.invoke("what is GenAI?")
print(response.content)
print(response.text)
```

```python
##using prompting method
import os
from langchain_groq import ChatGroq
from dotenv import load_dotenv

load_dotenv()

llm = ChatGroq(
    model="openai/gpt-oss-120b",
    api_key=os.getenv("GROQ_API_KEY"),
    temperature=0
)
#static pattern
prompt=[
    ("system","You are a python developer"),
    ("user","tell me somthing about list"),
]
response=llm.invoke(prompt)
print(response.content)
```
