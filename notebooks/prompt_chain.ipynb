```python
import os
from langchain_groq import ChatGroq
from dotenv import load_dotenv
from langchain_core.prompts import ChatPromptTemplate

load_dotenv()

llm = ChatGroq(
    model="openai/gpt-oss-120b",
    api_key=os.getenv("GROQ_API_KEY"),
    temperature=0
)
prompt = ChatPromptTemplate.from_messages([
    {"role":"system", "content":"You are a professional translator. Translate the given text into {language}."},
    {"role":"human","content": "{text}"}
])

final_prompt=prompt.format_messages(language="hindi",text="i love you")
response=llm.invoke(final_prompt)
print(response.content)
```

```python
#prompt using chaiming

import os
from langchain_groq import ChatGroq
from dotenv import load_dotenv
from langchain_core.prompts import ChatPromptTemplate

load_dotenv()

llm = ChatGroq(
    model="openai/gpt-oss-120b",
    api_key=os.getenv("GROQ_API_KEY"),
    temperature=0
)

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a professional translator. Translate the given text into {language}."),
    ("human", "{text}")
])

input_dict = {
    "language": "Hindi",
    "text": "Artificial Intelligence is changing the world."
}

chain = prompt | llm

response = chain.invoke(input_dict)

print(response.content)
```
